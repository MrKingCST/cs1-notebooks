# Part 1: The Big Ideas

## What is Python?
**Python** is a **programming language** — a way to give instructions to a computer.

Computers are fast, but they don't understand English. Programming languages are a middle ground: written so *people* can read them, but strict enough that a computer can follow them exactly.

Python is one of the most popular languages in the world because it is **easy to read**. It's used by:
- Game developers
- Data scientists and sports analysts (stats, shot charts, player rankings)
- People building AI and robots
- Companies like Google, Netflix, Spotify, and NASA

## What is an IDE?
**IDE** stands for **Integrated Development Environment**. That's a fancy name for *a program that helps you write code.*

Think of it like a workshop. You *could* build a birdhouse with just a hammer, but a workshop puts every tool you need in one place. An IDE gives programmers:

| Tool | What it does |
|---|---|
| **Editor** | A place to type your code |
| **Run button** | Runs your code so you can see what happens |
| **Output area** | Shows the results of your code |
| **Error messages** | Tells you when something is wrong — and usually where |
| **Color highlighting** | Colors different parts of code so it's easier to read |

# Part 3: How `print()` Works

`print()` is a **function** — a command that does a job. Its job is to display something on the screen.

```python
print("Hello, World!")
```

Let's break it down:

| Part | Name | What it means |
|---|---|---|
| `print` | function name | "Display something" — must be **lowercase** |
| `( )` | parentheses | Hold whatever you want printed |
| `" "` | quotation marks | Show where the text starts and stops |
| `Hello, World!` | string | The text itself. In Python, text is called a **string** |

Run each cell below and look at the output. **Predict** what will happen *before* you run it!

### Example 1: Printing more than one line
Each `print()` starts on a new line.

````python
print("Welcome to Computer Science!")
print("Python is fun.")
print("Let's get coding.")
````

### Example 2: Single or double quotes
Both work! Just make sure they **match**.

````python
print('This uses single quotes.')
print("This uses double quotes.")
````

### Example 3: Printing numbers
Numbers **don't** need quotes.

````python
print(23)
print(3.14)
````

### Example 4: Python can do math
Without quotes, Python **calculates** the answer. With quotes, it just prints the text exactly as written.

**Predict:** What will each line print?

````python
print(10 + 5)
print("10 + 5")
print(6 * 7)
print(100 / 4)
````

| Symbol | Meaning |
|---|---|
| `+` | add |
| `-` | subtract |
| `*` | multiply |
| `/` | divide |

### Example 5: Printing several things at once
Use **commas** to print multiple items on one line. Python adds a space between them.

````python
print("My team scored", 58, "points!")
print("2 + 2 =", 2 + 2)
````

### Example 6: A blank line
An empty `print()` prints a blank line — handy for spacing out your output.

````python
print("Line one")
print()
print("Line three (with a gap above)")
````

### Example 7: Comments
A line that starts with `#` is a **comment**. Python **ignores** it. Programmers use comments to leave notes for people reading the code.

````python
# This is a comment. Python skips this line.
print("Only this line shows up!")  # comments can go at the end of a line, too
````
