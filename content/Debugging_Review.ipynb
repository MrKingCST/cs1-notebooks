# Part 1: How to Read an Error

When something goes wrong, Python prints a red error message. It can look scary, but you only need two things from it:

1. **The last line.** It names the type of error and usually gives a clue.
2. **The line number or arrow** (`---->` or `^`). It points to where Python got stuck.

Here are the errors you've seen so far:

| Error | What it usually means | Example |
|---|---|---|
| `SyntaxError` | Python can't read the line. Look for a missing quote, parenthesis, or colon. | `print("Hi)` |
| `NameError` | Python doesn't know that name. Check spelling and capital letters. | `Print("Hi")` |
| `TypeError` | You mixed types that don't go together, like text and a number. | `"Age: " + 14` |
| `ValueError` | The type is right, but the value won't work. | `int("2.5")` |
| `IndentationError` | A line is missing its indent, or has one it shouldn't. | A line under `if` that isn't indented |
| `IndexError` | You asked for a list position that doesn't exist. | `roster[10]` |

Two kinds of bugs **don't** give an error message at all:

* A **logic error**: the code runs, but the answer is wrong.
* An **infinite loop**: the cell never stops. Click the stop button (the square on the left of the cell, where ▶️ usually is) to stop it.

**Tip for logic errors:** add a `print()` to see what a variable holds at that moment. Delete it when you're done.

# Final Round: The Big Bug Hunt

This program has **4 bugs** from different lessons.

**Goal:** It should print the team's average points per game, ask for the team's goal, and then say whether the team is meeting it.

⚠️ **Python only shows you one error at a time.** It stops at the first problem it finds. Fix that one, run it again, and keep going until all 4 are fixed.

When it works, test it twice:

* Type `50` for the goal. It should say the team is meeting its goal.
* Type `60` for the goal. It should say the team needs to score more.

```python
# Team Stats Program
scores = [58, 49, 62, 55]
total = 0

for score in scores
    total = total + score

average = total / len(scores)
print("Average points: " + average)

goal = input("What is the team's goal for points per game? ")

if average >= goal:
    print("The team is meeting its goal!")
else:
print("The team needs to score more.")
```

<details>
<summary><b>Click here after you've fixed all 4</b> to check your thinking</summary>

1. **SyntaxError: expected ':'** on the `for` line. Add a colon: `for score in scores:`
2. **IndentationError** under `else:`. Indent the last `print` line.
3. **TypeError: can only concatenate str (not "float") to str.** Use a comma: `print("Average points:", average)`
4. **TypeError: '>=' not supported between instances of 'float' and 'str'.** `input()` gave back a string, so Python can't compare it to a number. Fix: `goal = float(input(...))` (or `int()`)

When it's fixed, it prints `Average points: 56.0`.

Notice that Python found the two "can't read it" errors (1 and 2) before it ran any code. It only found errors 3 and 4 while the program was running.
</details>
