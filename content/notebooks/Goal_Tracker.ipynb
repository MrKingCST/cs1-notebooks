⚠️ An `if` statement only checks **once**. If the user makes the same mistake twice, an `if` will let the second bad answer through. Which kind of loop keeps checking **until** something changes?

**Check before you move on:** Type `0` for the number of days, then `0` again, then `-5`. Your program should keep asking every time.

<details>
<summary>Hint 1: How the check works</summary>

1. Ask the question like normal.
2. Start a `while` loop that keeps going as long as the answer is bad.
3. Inside the loop, print a message and ask the question **again**, storing the new answer in the **same** variable.

When the user finally types a good answer, the `while` condition becomes `False` and the program moves on.
</details>

<details>
<summary>Hint 2: Where do the goal and actual checks go?</summary>

Each check goes **inside** your `for` loop, right after you ask for that number and **before** you `append()` it. That means the code inside the `while` loop will be indented twice.
