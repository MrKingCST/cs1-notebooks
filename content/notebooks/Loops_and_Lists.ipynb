# Part 2: Lists

So far, each variable has held **one** value. But a team has more than one player. Making a separate variable for every player (`player1`, `player2`, `player3`...) would get messy fast.

A **list** stores many values in one variable. Put the items inside square brackets `[ ]` and separate them with commas.

````python
roster = ["Jordan", "Ava", "Marcus", "Eli"]
scores = [58, 49, 62]

print(roster)
print(scores)
````

Notice that when you print a whole list, Python shows the brackets and the quotes too.

## Getting one item: the index

Every item in a list has a position number called an **index**.

⚠️ **Python starts counting at 0, not 1.** The first item is at index `0`.

| Index | 0 | 1 | 2 | 3 |
|---|---|---|---|---|
| Item | `"Jordan"` | `"Ava"` | `"Marcus"` | `"Eli"` |

To get one item, put its index in square brackets after the list's name.

**Predict:** What will each line print?

````python
roster = ["Jordan", "Ava", "Marcus", "Eli"]

print(roster[0])
print(roster[2])
print(len(roster))
````

<details>
<summary>Click after you've run it</summary>

1. `Jordan` (index 0 is the first item)
2. `Marcus` (index 2 is the **third** item)
3. `4` (`len()` tells you how many items are in the list)

The list has 4 items, but the last index is 3. The last index is always one less than the length.
</details>

## Adding an item: append()

`append()` adds a new item to the **end** of a list.

````python
roster = ["Jordan", "Ava", "Marcus", "Eli"]
roster.append("Sam")

print(roster)
print(len(roster))
````

Notice the dot: `roster.append("Sam")` means "add `"Sam"` to the end of `roster`."

# Part 3: for Loops

A **loop** repeats code so you don't have to type it over and over.

A `for` loop runs its indented code **once for each item** in a list.

````python
roster = ["Jordan", "Ava", "Marcus", "Eli"]

for player in roster:
    print("Good game,", player)

print("Team meeting is over.")
````

Let's break it down:

| Part | What it means |
|---|---|
| `for` | Starts the loop |
| `player` | A variable that holds **one item at a time**. The first time through it's `"Jordan"`, then `"Ava"`, and so on. You can name it anything. |
| `in roster` | The list to go through |
| `:` | Ends the line, just like with `if` |
| the indented line | Runs once for every item |
| `print("Team meeting is over.")` | Not indented, so it runs **once**, after the loop is done |

## Looping a set number of times: range()

`range()` gives a loop a list of numbers to count through.

**Predict:** What numbers will each loop print?

````python
for number in range(5):
    print(number)

print("---")

for number in range(1, 6):
    print(number)
````

<details>
<summary>Click after you've run it</summary>

* `range(5)` gives `0, 1, 2, 3, 4`. It starts at 0 (just like an index) and gives you 5 numbers.
* `range(1, 6)` gives `1, 2, 3, 4, 5`. It starts at the first number and **stops before** the second number.

To count from 1 to 10, use `range(1, 11)`.
</details>

## Adding up a list: a running total

Loops are great for adding up numbers. Start a variable at `0`, then add each item to it.

````python
scores = [58, 49, 62]
total = 0

for score in scores:
    total = total + score

print("Total points:", total)
````

Here's what happens to `total` each time through the loop:

| Time through the loop | `score` | `total` after this time |
|---|---|---|
| Before the loop | | `0` |
| 1st | `58` | `58` |
| 2nd | `49` | `107` |
| 3rd | `62` | `169` |

The `print` is not indented, so it only runs once, after the loop finishes.

# Part 4: while Loops

A `while` loop repeats **as long as its condition is `True`**. It's like an `if` statement that keeps checking over and over.

````python
lap = 1

while lap <= 5:
    print("Lap", lap)
    lap = lap + 1

print("Done running!")
````

How it works:

1. Python checks the condition: is `lap <= 5`?
2. If `True`, it runs the indented code, then goes back to step 1.
3. If `False`, it skips the loop and moves on.

The line `lap = lap + 1` is very important. Without it, `lap` would stay `1` forever, the condition would never become `False`, and the loop would **never stop**. That's called an **infinite loop**.

⚠️ **If a cell keeps running and won't stop,** click the stop button (the square on the left of the cell, where ▶️ usually is). You'll see a `KeyboardInterrupt` message. That's normal. It just means you stopped it.

## Repeating until the user gets it right

A `while` loop is perfect when you don't know how many times something will repeat. This one keeps asking until the user types `yes`.

````python
answer = ""

while answer != "yes":
    answer = input("Ready for tip-off? Type yes: ")

print("Let's go!")
````

Why does it start with `answer = ""`? The `while` line checks `answer` right away, so `answer` has to exist before the loop starts. `""` is an empty string: text with nothing in it.

**Try it:** Run the cell and type something other than `yes` a few times before typing `yes`.

⚠️ Type `yes` in all lowercase. Python treats `Yes` and `yes` as different words, so `Yes` won't end the loop.

## for or while?

| Use a... | When... | Example |
|---|---|---|
| `for` loop | You have a list, or you know how many times to repeat | Print every player on the roster |
| `while` loop | You repeat **until** something changes | Keep asking until the password is right |
