# Part 1: Variables

In Lesson 1, every time we wanted to print something, we had to type it out. What if we want to **save** a piece of information and use it again later?

That's what a **variable** is for.

### What is a variable?

A **variable** is a name that stores a value.

Think of a variable like a **labeled box**. You put something inside the box, write a label on the outside, and later you can grab what's inside just by reading the label.

````python
player = "Jordan"
print(player)
````

Let's break it down:

| Part | Name | What it means |
|---|---|---|
| `player` | variable name | The label on the box |
| `=` | assignment operator | "Store the value on the right in the box on the left" |
| `"Jordan"` | value | What goes inside the box |

⚠️ **Important:** In Python, `=` does **not** mean "equals" like in math class. It means **"store this."** Read `player = "Jordan"` as *"player gets Jordan."*

Notice: when we printed, we wrote `print(player)` with **no quotes**. That tells Python to look inside the box.

### Example 1: Quotes vs. no quotes

**Predict:** What will each line print?

````python
mascot = "Patriots"
print(mascot)
print("mascot")
````

- `print(mascot)` → no quotes, so Python looks inside the variable and prints **Patriots**.
- `print("mascot")` → quotes, so Python prints the word **mascot** exactly as written.

This is the same idea as `print(5 + 5)` vs. `print("5 + 5")` from Lesson 1!

### Example 2: Variables can hold numbers, too

````python
points = 18
rebounds = 7
free_throw_pct = 0.85

print(points)
print(rebounds)
print(free_throw_pct)
````

### Example 3: Doing math with variables

Once a number is in a variable, you can use it in math — just like a number.

````python
first_half = 24
second_half = 31
total = first_half + second_half

print("Total points:", total)
````

### Example 4: Variables can change

That's why they're called *variables* — their value can **vary**. When you store a new value, the old one is replaced.

**Predict:** What will each `print()` show?

````python
score = 0
print("Start of game:", score)

score = score + 2
print("After a layup:", score)

score = score + 3
print("After a 3-pointer:", score)
````

`score = score + 2` means: *"Take what's in `score`, add 2, and store the answer back in `score`."*

### Example 5: Data types

Every value in Python has a **type**. Here are the three we'll use the most:

| Type | Short for | Examples |
|---|---|---|
| `str` | string (text) | `"Jordan"`, `"hello"`, `"58"` |
| `int` | integer (whole number) | `18`, `0`, `-5` |
| `float` | decimal number | `3.14`, `0.85`, `25.0` |

The `type()` function tells you what type something is. Run this:

````python
print(type("Jordan"))
print(type(18))
print(type(0.85))
print(type("18"))
````

Look at the last one! `"18"` is a **string**, not a number, because it's in quotes. Remember this — it's going to matter in Part 3.

# Part 2: Naming Variables

You get to choose your variable names, but Python has some rules.

| Rule | ✅ Good | ❌ Bad |
|---|---|---|
| Use letters, numbers, and underscores `_` only | `player_2` | `player-2`, `player!` |
| No spaces — use `_` instead | `first_name` | `first name` |
| Can't **start** with a number | `team1` | `1team` |
| Can't be a Python word | `my_print` | `print` |

**Also remember:** variable names are **case-sensitive**. `Score`, `score`, and `SCORE` are three different variables!

### Good habits

- Use names that describe what's inside: `total_points` is better than `x`.
- Use lowercase with underscores between words: `free_throws_made`.

# Part 3: Getting Input from the User

So far, **we** have decided what goes in every variable. But real programs ask the **user** for information — like a game asking for your username.

The `input()` function asks a question, waits for the user to type an answer, and gives that answer back so you can store it in a variable.

### Example 6: Your first input

Run the cell. A box will appear — type your name and press **Enter**.

````python
name = input("What is your name? ")
print("Nice to meet you,", name)
````

Let's break it down:

| Part | What it does |
|---|---|
| `input(...)` | Shows the question and waits for the user to type |
| `"What is your name? "` | The question (called a **prompt**). Add a space at the end so the answer isn't squished against the question! |
| `name = ` | Stores whatever the user typed in the variable `name` |

### Example 7: Asking more than one question

````python
name = input("What is your name? ")
food = input("What is your favorite food? ")

print()
print(name, "loves", food + "!")
````

### Example 8: The input trap

Run this cell and type your age. **It's going to crash** — that's on purpose! Read the last line of the error.

**Note** that "concatenate" means to join together.

````python
age = input("How old are you? ")
print("Next year you will be", age + 1)
````

### What happened?

`input()` **always** gives back a **string** — even if you type a number. So if you type `14`, Python stores `"14"` (text), not `14` (a number).

Python can't add a number to text, so it gives a **TypeError**.

### The fix: `int()` and `float()`

We can **convert** a string into a number:

| Function | Converts to | Example |
|---|---|---|
| `int()` | whole number | `int("14")` → `14` |
| `float()` | decimal number | `float("5.5")` → `5.5` |

````python
age = int(input("How old are you? "))
print("Next year you will be", age + 1)
````

Now it works! Wrapping `input()` inside `int()` turns the answer into a number right away.

**Rule of thumb:** If you plan to do **math** with the answer, wrap the `input()` in `int()` or `float()`.

### Example 9: Putting it all together

````python
player = input("Player name: ")
twos = int(input("How many 2-pointers did they make? "))
threes = int(input("How many 3-pointers did they make? "))

total = twos * 2 + threes * 3

print()
print(player, "scored", total, "points!")
````
