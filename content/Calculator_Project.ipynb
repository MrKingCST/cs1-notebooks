# Part 2: Data Types

Every value in Python has a **data type**. The type tells Python what kind of value it is, and what it's allowed to do with it.

Here are the three types you need for a calculator:

| Type | Python name | What it holds | Examples |
|---|---|---|---|
| String | `str` | Text, written inside quotes in your code | `"hello"`, `"Jordan"`, `"14"` |
| Integer | `int` | Whole numbers | `7`, `-3`, `100` |
| Float | `float` | Numbers with a decimal point | `3.14`, `0.5`, `2.0` |

Notice that `"14"` is a **string**, not a number. If it's in quotes, it's text, even if it looks like a number.

### Checking a type with `type()`

`type()` tells you what type a value is.

**Predict:** Before you run this, write down what type you think each one will be.

````python
print(type("hello"))
print(type(14))
print(type(14.0))
print(type("14"))
````

### Why does the type matter?

The same symbol can do **different jobs** depending on the type.

**Predict:** What will each line print?

````python
print(5 + 3)
print("5" + "3")
print(3 * 2)
print("ha" * 3)
````

With numbers, `+` **adds**. With strings, `+` **glues them together**. That's called **concatenation**. `"5" + "3"` becomes `"53"`, the same way `"basket" + "ball"` becomes `"basketball"`.

With strings, `*` **repeats** the text.

### Wait... 53?!

Here's the secret: **`input()` always gives back a string**, even if the user types a number. So Python saw `"5" + "3"` and glued them together.

To do real math with what the user types, we have to **convert** it into a number first.

### Mixing ints and floats

Ints and floats **can** be used together in math. Run this and look closely at the answers:

````python
print(4 + 2.5)
print(10 / 2)
print(10 * 2)
print(type(10 / 2))
````

Two rules to notice:

1. If **either** number is a float, the answer is a float.
2. Division with `/` **always** gives a float, even when it divides evenly. That's why `10 / 2` is `5.0`, not `5`.

# Part 3: Converting Between Types

Python has a function for each type that converts a value **into** that type:

| Function | Converts to | Example | Result |
|---|---|---|---|
| `int()` | Integer | `int("14")` | `14` |
| `float()` | Float | `float("2.5")` | `2.5` |
| `str()` | String | `str(14)` | `"14"` |

To convert what a user types, wrap `input()` inside the conversion function:

````python
age = int(input("How old are you? "))
print("Next year you will be", age + 1)
````

Look closely at the parentheses: `int(input("How old are you? "))`

Python works from the **inside out**:

1. `input(...)` runs first and gives back a string, like `"14"`
2. `int(...)` turns `"14"` into the number `14`
3. `age =` stores `14` in the variable

That's why there are **two** `)` at the end: one closes `input(`, one closes `int(`.

### Predict the conversion

Conversions don't always do what you expect. Predict each output, then run it.

````python
print(int(9.99))
print(float(7))
print(str(5) + str(5))
print(int("5") + int("5"))
````

<details>
<summary>Click after you've run it</summary>

* `int(9.99)` gives `9`. `int()` **chops off** the decimal. It does **not** round!
* `float(7)` gives `7.0`
* `str(5) + str(5)` gives `55` (two strings glued together)
* `int("5") + int("5")` gives `10` (two numbers added)
</details>
