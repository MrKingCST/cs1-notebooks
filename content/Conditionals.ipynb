# Part 2: True or False?

Every program you've written so far runs every line, top to bottom, every time. Real programs make **decisions**:

* A game checks: is the player's health 0? If so, show "Game Over."
* A scoreboard checks: is the home score higher? If so, the home team wins.

To make a decision, Python asks a **yes-or-no question** called a **condition**. The answer is always `True` or `False`. These two values are called **Booleans** (BOO-lee-uhns).

We ask these questions with **comparison operators**:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `>` | greater than | `10 > 7` | `True` |
| `<` | less than | `10 < 7` | `False` |
| `>=` | greater than or equal to | `5 >= 5` | `True` |
| `<=` | less than or equal to | `4 <= 3` | `False` |
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 5` | `False` |

⚠️ **Important:** `=` and `==` are different.

* `=` **stores** a value: `score = 10` means "put 10 in score."
* `==` **asks a question**: `score == 10` means "is score equal to 10?"

**Predict:** Write down `True` or `False` for each line, then run it.

````python
print(58 > 49)
print(12 < 3)
print(7 == 7.0)
print(20 != 21)
print("Patriots" == "Patriots")
print("Patriots" == "patriots")
````

# Part 3: The if Statement

In English we say, "**If** it's raining, **then** bring an umbrella."

Python works the same way, but it leaves out the word "then." The colon `:` and the indented line do that job.

````python
temperature = 95

if temperature > 90:
    print("It's hot! Practice is moving to the gym.")

print("Have a great day.")
````

Let's break it down:

| Part | What it means |
|---|---|
| `if` | Starts the question |
| `temperature > 90` | The condition. Python checks if it's `True` or `False` |
| `:` | Ends the question. Don't forget it! |
| the indented line | Only runs **if the condition is True** |
| `print("Have a great day.")` | Not indented, so it **always** runs |

⚠️ **Indentation matters in Python.** The indented lines "belong" to the `if`. Colab indents for you when you press Enter after a colon. If you need to do it yourself, press **Tab** or type 4 spaces.

# Part 4: else

`if` by itself only does something when the condition is `True`. Add `else` to do something different when it's `False`.

Run this cell twice. The first time, give the home team the higher score. The second time, give the away team the higher score.

````python
home = int(input("Home team score: "))
away = int(input("Away team score: "))

if home > away:
    print("Home team wins!")
else:
    print("Home team did not win.")
````

How `else` works:

* `else` has **no condition**. It runs whenever the `if` condition is `False`.
* `else` lines up with `if` (not indented), and it needs a colon too.
* Exactly **one** of the two blocks runs. Never both, never neither.

Notice the `int()` around each `input()`. Without it, Python would compare two strings instead of two numbers.

# Part 5: elif

When there are more than two possibilities, use `elif`. It is short for "**el**se **if**."

A game can end three ways: a win, a tie, or a loss.

````python
home = int(input("Home team score: "))
away = int(input("Away team score: "))

if home > away:
    print("Home team wins!")
elif home == away:
    print("It's a tie. Overtime!")
else:
    print("Home team lost. Get 'em next time.")
````

How Python reads an `if` / `elif` / `else` chain:

1. It checks the conditions **from top to bottom**.
2. It runs the block under the **first** condition that is `True`, then skips the rest of the chain.
3. If none of them are `True`, it runs the `else` block.

You can use as many `elif`s as you need. The `else` is optional, but it must always come **last**.

**Predict:** What will this print? Be careful. More than one condition is true!

````python
points = 22

if points >= 30:
    print("Huge night!")
elif points >= 20:
    print("Great game!")
elif points >= 10:
    print("Solid game.")
else:
    print("Keep working.")
````

<details>
<summary>Click after you've run it</summary>

It prints `Great game!`

`22 >= 10` is also `True`, but Python stops at the **first** true condition and skips the rest of the chain. That's why the order of your conditions matters.
</details>

## Bug #4

The score is 95, so this should print `A`.

````python
score = 95

if score >= 70:
    print("C")
elif score >= 80:
    print("B")
elif score >= 90:
    print("A")
else:
    print("Below a C")
````

<details>
<summary><b>Click here after you've tried all four</b> to check your thinking</summary>

1. **SyntaxError: expected ':'**. The `if` line needs a colon at the end: `if points > 20:`
2. **SyntaxError: invalid syntax**. Use `==` to compare: `if mascot == "Patriots":`. A single `=` stores a value.
3. **IndentationError**. The `print` line belongs to the `if`, so it must be indented.
4. **Logic error.** `95 >= 70` is `True`, so Python prints `C` and skips the rest. Put the highest score first: check `>= 90`, then `>= 80`, then `>= 70`.
